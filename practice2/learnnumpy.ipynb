## 1. Array/Matrix Creation

### 1.1 Converting Python sequences to NumPy Arrays
NumPy arrays can be defined using Python sequences such as lists and tuples. Lists and tuples are defined using [...] and (...), respectively. Lists and tuples can define ndarray creation:

- a list of numbers will create a 1D array,

- a list of lists will create a 2D array,

- further nested lists will create higher-dimensional arrays. In general, any array object is called an ndarray in NumPy.

```python
arr = np.array([0, 1, 2, 3, 4, 5])
# use a list to initialize a one-dimensional numpy array
arr
```

```python
type(arr)  # <class 'numpy.ndarray'>
# type() returns the type of the object inside the parentheses.
# in this case, it returns the type of arr, which is a numpy array.
```

```python
arr.itemsize
# itemsize returns the size of each element in the array in bytes.
# this time it returns 8,
# meaning that each element in the array is 8 bytes long.
```

```python
arr = np.array([0, 1, 2, 3, 4, 5], dtype = 'int16')
# dtype specifies the data type of the elements in the array, which is int16,
# meaning that each element in the array is a 16-bit integer.

arr.itemsize
# each element = 16 bits = 2 bytes -> itemsize returns 2
```

```python
arr.dtype
# dtype returns the data type of the elements in the array.
```

```python
arrf = arr.astype('float32')
arr,arrf
# astype() converts int16 -> float32
```

```python
# use a list of lists to initialize a two-dimentional numpy array
arr2d = np.array([[1, 2], [3, 4]])

#use further nested lists to initialize a higher-dimensional arrays
arr3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

arr2d,arr3d
```

### 1.2 Intrinsic NumPy array creation functions

#### 1D array creation functions
<hr>

**numpy.arange** creates arrays with regularly incrementing values.
<https://numpy.org/doc/stable/reference/generated/numpy.arange.html#numpy.arange>

```python
np.arange(10)

np.arange(10)
# arange = array + range
# arrange(10) returns an array of integers from 0 to 9.
```

```python
np.arange(2, 10, dtype=float)
# [2, 10) -> start from 2, end at 10 (but exclusive)
# float -> 2., but not 2
```

Note: best practice for numpy.arange is to use integer start, end, and step values. There are some subtleties regarding dtype. In the second example, the dtype is defined. In the third example, the array is dtype=float to accommodate the step size of 0.1. Due to roundoff error, the stop value is sometimes included.

**numpy.linspace** will create arrays with a specified number of elements, and spaced equally between the specified beginning and end values.
<https://numpy.org/doc/stable/reference/generated/numpy.linspace.html#numpy.linspace>

```python
np.linspace(1., 4., 6)
# linspace() returns evenly-spaced numbers over a specified interval,
# with each adjacent number separated by a constant step size.
# in this case, the step size = (4 - 1) / 6 - 1 = 0.6
```

#### 2D array creation functions

<hr>

**numpy.diag** can define either a square 2D array with given values along the diagonal or if given a 2D array returns a 1D array that is only the diagonal elements.

```python
np.diag([1, 2, 3, 4])
# diag = diagonal
# it returns a square 2D array with elements of input on the diagonal, and 0s elsewhere.
```

```python
np.identity(4)# identity matrix
# identity() returns a square 2D array,
# with 1s on the main diagonal and 0s elsewhere.
```

```python
np.eye(3)
# eye() vs identity()
# eye() returns a 2D array with 1s on the diagonal and 0s elsewhere,
# but it allows you to specify the number of rows and columns,
# while identity() always returns a square 2D array with the same number of rows and columns.
```

#### General ndarray creation functions
The ndarray creation functions e.g. **numpy.ones**, **numpy.zeros**, and **numpy.full** define arrays based upon the desired shape. The ndarray creation functions can create arrays with any dimension by specifying how many dimensions and length along that dimension in a tuple or list.

```python
np.zeros((10,10))
# zeros() returns a 2D array of the specified shape, filled with 0s.
# in this case, the shape: 10 rows * 10 cols.
```

#### Replicating, joining, or mutating existing arrays
Once you have created arrays, you can replicate, join, or mutate those existing arrays to create new arrays. When you assign an array or its elements to a new variable, you have to explicitly **numpy.copy** the array, otherwise the variable is a view into the original array. 

```python
a = np.array([1, 2, 3, 4, 5, 6])
b = a[:2] # b is a view of the first two elements of a, not a copy.

b += 1
# This modifies the first two elements of a as well, since b is a view of a.
# a and b point to the same data in memory, so changing b also changes a.
# += 1 -> increment the value of each element in b by 1

a,b
```

In above example, you did not create a new array. You created a variable, **b** that viewed the first 2 elements of a. When you added 1 to b you would get the same result by adding 1 to a[:2]. If you want to create a new array, use the numpy.copy array creation routine as such:

```python
a = np.array([1, 2, 3, 4, 5, 6])
b = a[:2].copy()
# now, different from the previous example,
# b is a copy of the first two elements of a, not a view,
# meaning that b now points to a different location in memory than a,
# so this time changing b will not affect a.

b += 1
a,b
```

**numpy.where**  
numpy.where(condition, [x, y, ]/)  
return an array with elements from x where condition is True, and elements from y elsewhere.

```python
x, y = np.ogrid[:3, :4]
# ogrid() returns an open meshgrid, which is a set of arrays that represent the coordinates of a grid in a multi-dimensional space.
# this case, x = [[0], [1], [2]], y = [[0, 1, 2, 3]]
# :3 means 0, 1, 2 (3 is exclusive), :4 means 0, 1, 2, 3 (4 is exclusive)

print(x, y)
z = np.where(x < y, x, 10 + y)
# (x < y) is the condition, if it is true, return x, otherwise return 10 + y
z
```

```python
np.where(z < 10, 0, 10)
# condition: z < 10, if true -> 0, otherwise -> 10
```

**numpy.sort(a, axis=- 1, kind=None, order=None)**

Return a sorted copy of an array.

```python
x = np.array([5, 4, 6, 8, 1, 2, 9])
print('x:', x)
x.sort()
print('x:', x)
# sort in ascending order
```

```python
x = np.array([[7, 5, 8], [9, 6, 4], [3, 4, 9]])
print('x:\n', x)
x.sort(axis=0)
print('x:\n', x)
```

**numpy.argsort(a, axis=- 1, kind=None, order=None)**

Returns the indices that would sort an array.

```python
x = np.array([5, 4, 6, 8, 1, 2, 9])
print('arr:\n', x)
x_index = np.argsort(x)
print('x:\n', x)
print('x_index:\n', x_index)
```

```python
x = np.array([[7, 5, 8], [9, 6, 4], [3, 4, 9]])
print('x:\n', x)
x_index = np.argsort(x, axis=-1)
# argsort() returns the indices that would sort an array along the specified axis.
# axis=-1 means that the sorting is done along the last axis (columns) of the array.
print('x:\n', x)
# x remains unchanged
print('x_index:\n', x_index)
# but x_index contains the indices that would sort each row of x in ascending order.
# e.g. [7, 5, 8] -> [5, 7, 8]
# 7's index = 1, 5's index = 0, 8's index = 2
```
